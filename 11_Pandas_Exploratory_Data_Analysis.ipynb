#4. Outlier Detection
What is an Outlier?

An outlier is a data point that lies far away from other values in a dataset.

Why it matters:

- Outliers can skew your - analysis (mean, variance)
- Sometimes they are errors, sometimes true extreme events
- You need to detect and decide what to do

Boxplot Basics

A boxplot shows:

1. Q1 (25%) → lower quartile
2. Q2 (50%) / Median → middle value
3. Q3 (75%) → upper quartile
4. IQR (Interquartile Range) = Q3 − Q1
5. Whiskers → extend to Q1 − 1.5*IQR and Q3 + 1.5*IQR
6. Outliers → points outside whiskers

##Step 1 — Plotting a Boxplot

We will use Matplotlib first:

```python
plt.boxplot(dff['With_Outliers'])
plt.title("Boxplot for Sales")
plt.ylabel("Sales")
plt.show()
```

```python
dff['With_Outliers'].describe()
```

##Step 2 — Using Seaborn (more elegant)

```python
import seaborn as sns

sns.boxplot(y=dff['With_Outliers'])
plt.title("Boxplot for Sales")
plt.show()
```

##Step 3 — Identifying Outliers

We can calculate them programmatically:

```python
Q1 = dff['With_Outliers'].quantile(0.25)
Q3 = dff['With_Outliers'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5*IQR
upper_bound = Q3 + 1.5*IQR

outliers = dff[(dff['With_Outliers'] < lower_bound) | (dff['With_Outliers'] > upper_bound)]
outliers
```

##PART 1 — NUMERIC vs NUMERIC
1. Correlation (df.corr())

```python
df.corr(numeric_only=True) #Sales Dataset
```

```python
df_cat.corr(numeric_only=True)
```

###What is Correlation?

Measures relationship strength between two variables

## Range:
|Value|	Meaning|
|-----|-----------
|+1|	Strong positive|
|0|	No |relationship|
|-1|	Strong negative|

###Interpretation
####Positive correlation

- One increases → other increases

####Negative correlation

- One increases → other decreases

```python
df_cat[['Sales', 'Profit']].corr()
```

Sales and Profit show a strong positive correlation, indicating that higher sales are associated with higher profits.

#2. Scatter Plot

```python
plt.scatter(df_cat['Sales'], df_cat['Profit'])

plt.xlabel("Sales")
plt.ylabel("Profit")
plt.title("Sales vs Profit")

plt.show()
```

##How to Read Scatter Plot
- Upward trend → Positive correlation
- Downward trend → Negative correlation
- Random → No relationship

The scatter plot shows a clear upward trend, confirming a strong positive relationship between Sales and Profit.
