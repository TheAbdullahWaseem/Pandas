#B. Pandas Series
A Series is a **1-dimensional labeled array** capable of holding any data type **(integers, floats, strings, etc.).**

**Each element has an index, which is like a row label.**

**Think of Series as a single column of a DataFrame.**

**Vectorized operations work on Series.**

##Creating Series 1. From a list

```python
data = [10,20,30,40]
s = pd.Series(data)
print(s)
```

##2. From a list with custom index

```python
s1 = pd.Series(data, index = ['a','b','c','d'])
print(s1)
```

##3. From a Dictionary

```python
data_dict = {'Math':90,'Science': 85, 'English': 95}
s2 = pd.Series(data_dict)
print(s2)
```

##4. Check values, index, dtype

```python
print("Values of s2 and s1 are: \n")
print(s2.values)
print(s1.values, "\n")

print("Index of s2 and s1 are: \n")
print(s2.index)
print(s1.index, "\n")

print("dtype of s2 and s1 are: \n")
print(s2.dtype)
print(s1.dtype)
```

#Series Operations are Vectorized

```python
print(s)
```

##Multiply whole column with 5, Vectorized

```python
print(s*5)
```

```python
print(s2*5)
```

#Can select elements by index or position:

##Select Element by Position

```python
print(s[1])
```

##Select element by Label

```python
print(s2['Math'])
```

##Missing values are represented as Nan

```python
missing = [10, None, 50, None, 70, 80, None]
s3 = pd.Series(missing, index = ['a','b','c','d','e','f','g'])
print(s3)
```

#C. Pandas DataFrame
A DataFrame is a **2-dimensional labeled** data structure, like an Excel spreadsheet or SQL table.

**Consists of rows and columns**.

**Each column is a Series, but columns can have different data types.**

**Index identifies rows; column names identify columns.**

**All pandas operations (filtering, cleaning, grouping) work on DataFrames**

##Key properties:

df.shape → rows x columns

df.columns → column names

df.index → row labels

df.dtypes → column data types

##Creating DataFrames
###1. From Dictionary

```python
data = {
    'Name': ['Ali', 'Sana', 'Omar'],
    'Math': [90, 80, 75],
    'Science': [85, 95, 65]
}
df = pd.DataFrame(data)
print(df)
```

###If you add index to dictionaries columns names changed to 0,1,2...

```python
df3 = pd.DataFrame(data, index = ['a','b','c'])
print(df3)
```

```python
df3 = pd.DataFrame(data, index = ['a','b','c'], columns = ['Math', "Science", "Name"])
print(df3)
```

##2. From list of lists with column names

```python
data = [
    ['Ali', 90, 85],
    ['Sana', 80, 95],
    ['Omar', 75, 65]
]
df1 = pd.DataFrame(data)
print(df1)
```

###Setting Column Names

```python
df2_names = pd.DataFrame(data, columns= ["Names", "Math", "English"])
print(df2_names)
```

###Setting Indexes

```python
df2_index = pd.DataFrame(data,index = ['a','b','c'])
print(df2_index)
```

###Setting Both Columns and Indexes

```python
df4 = pd.DataFrame(data, index= ['a','b','c'], columns= ['Name', 'Math', 'Science'])
print(df4)
```
