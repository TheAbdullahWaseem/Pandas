#6. Pandas Data Transformation Practice

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "employee": ["Ali", "Sara", "Omar", "Ayesha", "Bilal"],
    "base_salary": [5000, 4200, 6100, 4800, 5300],
    "bonus": [500, 300, 700, 400, 600],
    "tax_rate": [0.10, 0.08, 0.12, 0.09, 0.11],
    "working_days": [22, 20, 24, 21, 23]
})

df
```

#1 Creating New Columns
#2 Column Arithmetic Operations

##Problem 1 — Total Income

####Create a new column called total_income.

```python
df['total_income'] = df['base_salary'] + df['bonus']
df['total_income']
```

##Problem 2 — Monthly Tax Amount

####Create a column called tax_amount.

```python
df['tax_amount'] = df['base_salary'] * df['tax_rate']
df['tax_amount']
```

##Problem 3 — Net Salary After Tax

####Create a column net_salary.

```python
df['net_salary'] = (df['base_salary'] + df['bonus']) - df['tax_amount']
df['net_salary']
```

##Problem 4 — Salary Per Working Day

####Create a column called salary_per_day.

```python
df['salary_per_day'] = df['base_salary'] / df['working_days']
df
```

##Problem 5 — Bonus Percentage

####Create a column called bonus_percent.

```python
df['bonus_percent'] = (df['bonus']/df['base_salary']) * 100
df['bonus_percent']
```

#3. Apply() Function

```python
df = pd.DataFrame({
    "employee": ["Ali", "Sara", "Omar", "Ayesha", "Bilal"],
    "salary": [5000, 4200, 6100, 4800, 5300],
    "bonus": [500, 300, 700, 400, 600],
    "experience": [2, 4, 6, 3, 5]
})
df
```

##Problem 1

####Create a new column called salary_tax.
####You must use apply().

```python
df['salary_tax'] = df['salary'].apply(lambda x: x*0.10)
df['salary_tax']
```

##Problem 2 (Column Apply with Condition)

####Create a column bonus_after_cut.

###Rule:

####If bonus > 500 → reduce bonus by 50
####Otherwise keep bonus same

####You must use apply().

```python
df['bonus_after_cut'] = df['bonus'].apply(lambda x: x - 50 if x > 500 else x)
df['bonus_after_cut']
```

##Problem 3

####Create a column called total_income.

####But this time you must use row-wise apply.

```python
df['total_income'] = df.apply(lambda x: (x['salary'] + x['bonus']),axis= 1)
df['total_income']
```

##Problem 4 (Row Apply — Conditional Logic)

####Create a column experience_level.

###Rules:

####experience >= 5 → "Senior"
####experience >= 3 → "Mid"
####otherwise → "Junior"

###You must use apply().

```python
def ex(x):
  if x>=5:
    return "Senior"
  elif x>=3:
    return "Mid"
  else:
    return "Junior"


df['experience_level'] = df['experience'].apply(ex)
df[['experience', 'experience_level']]
```

##Problem 5

####Create a column called performance_score.

```python
def performance(x):
  score= x["salary"]*0.3 + x['bonus']*0.5 + x['experience']*100
  if score > 5000:
    return "Excellent"
  elif score > 4000:
    return "Good"
  else:
    return "Average"

df['performance_score'] = df.apply(performance, axis=1)
df['performance_score']
```

#4. Map() Function

```python
df = pd.DataFrame({
    "employee": ["Ali", "Sara", "Omar", "Ayesha", "Bilal"],
    "department": ["IT", "HR", "IT", "Finance", "HR"],
    "city": ["London", "Paris", "London", "Berlin", "Paris"],
    "experience": [2, 4, 6, 3, 5]
})
df
```

##Problem 1

###Convert department names to codes.

```python
df['department_codes'] = df['department'].map({"IT":101, "HR":102, "Finance": 103})
df['department_codes']
```

##Problem 2 — City Abbreviation Mapping

###Convert city names to short codes.

####London → LDN
####Paris → PAR
####Berlin → BER


```python
df['city_code'] = df['city'].map({"London":"LDN", "Paris":"PAR","Berlin":"BER"})
df['city_code']
```

##Problem 3 — Experience Category (Using Lambda)

###Create column:

####experience_level

###Rules:

####experience >= 5 → Senior
####otherwise → Junior

```python
df['experience_level'] = df['experience'].map(lambda x: "Senior" if x >= 5 else "Junior")
df
```

##Problem 4 — Name Length

###Create column:

####name_length

####It should contain the number of characters in the employee name.

```python
df['name_length'] = df['employee'].map(lambda x: len(x) )
df
```

##Problem 5 — Capitalize Names

###Create column:

####employee_upper

####Convert names to uppercase using map().

```python
df['employee_upper'] = df['employee'].map(lambda x: x.upper())
df
```

##Problem 6

####Some departments get priority scores.

###Mapping:

####IT → 3
####Finance → 2
####HR → 1

###Create column:

####priority_score

```python
df['priority_score'] = df['department'].map({'IT':3,"Finance":2, "HR":1}).fillna(0)
df
```
