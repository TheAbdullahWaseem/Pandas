#1. Create a Series of your 5 favorite movies with custom indices.

```python
movies = ["Top Gun Maverick", "A Beautiful Mind", "The Revenant", "Inception", "Interstellar"]
df = pd.Series(movies, index = ['1st', '2nd', '3rd', '4th', '5th'])
print(df)
```

#2. Create a DataFrame of 5 students and their scores in 3 subjects.

```python
students = {"Name": ["Ali", "Amir", "Asjid", "Fahad", "Hussain"],
            "Maths": [95, 90, 89, 92, 96],
            "Science": [90, 87, 90, 85, 87],
            "English": [89, 84, 89, 82, 88]}
df1 = pd.DataFrame(students)
print(df1)
```

#3. Inspect the DataFrame using head(), tail(), info(), describe().

```python
students = {"Name": ["Ali", "Amir", "Asjid", "Fahad", "Hussain"],
            "Maths": [95, 90, 89, 92, 96],
            "Science": [90, 87, 90, 85, 87],
            "English": [89, 84, 89, 82, 88],
            "Data Science": [98,87,98,95,89],
            "Machine Learning": [92,93,91,89,90]}
df2 = pd.DataFrame(students)
print(df2)
```

```python
print(df2.head())
```

```python
print(df2.tail())
```

```python
print(df2.info())
```

```python
print(df2.describe())
```

#4. Convert one numeric column to float and observe the change.

```python
df2["English"] = df2["English"].astype(float)
print(df2)
```

#5. Add a new column “Total” which sums the scores.

```python
df2['Total'] = df2[['Maths','Science','English','Data Science','Machine Learning']].sum(axis=1)
print(df2)
```

#6. Rename a column and change the index labels to custom names.

```python
df2.rename(columns= {"Maths":"Mathematics", "Science":"Scientific Studies"},inplace =True)
df2.index = ['1st','2nd','3rd','4th','5th']
print(df2)
```
