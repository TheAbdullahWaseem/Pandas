#2. Pandas Data Selection and Indexing
In this notebook, we explore data selection and indexing in Pandas, which are essential skills for extracting and filtering specific information from a dataset.

The following key concepts are covered:

**Selecting Columns** – Accessing one or multiple columns from a DataFrame using bracket notation.

**Selecting Rows** – Retrieving rows using index labels or positions.

**loc Indexer** – Label-based selection used to access rows and columns by their names or labels.

**iloc Indexer** – Position-based selection used to access rows and columns by integer index.

**Slicing Data** – Selecting ranges of rows or columns using slicing syntax.

**Boolean Filtering** – Filtering rows based on logical conditions (e.g., salary greater than a certain value).

**Conditional Filtering** – Combining multiple conditions using logical operators such as & (AND) and | (OR).

**Querying Data** – Using the query() method to filter data using readable conditional expressions.

```python
import pandas as pd
```

#D. Boolean Filtering
It selects rows based on conditions.

```python
df['Salary']>70000
```

##Filtering Data

```python
df[df['Salary']>70000]
```

##Multiple Conditions

```python
df[(df['Salary']>65000) & (df['Salary']<=80000)]
```

##And Condition

```python
df[(df['Salary']>70000) & (df['Department'] == 'Finance')]
```

##OR Condition

```python
df[(df['Salary']<70000) | (df['Department'] == 'IT')]
```

#E. Conditional Filtering
###We can combine conditions.

```python
df
```

```python
df[(df['Age']>24) & (df['Salary']>70000)]
```

#Querying Data with Query
###Another powerful filtering method.

```python
df.query("Salary>70000")
```

##Multiple conditions with query

```python
df.query("Age > 24 and Salary > 70000")
```

```python
df[(df["Age"]>24)&(df["Salary"]>70000)]
```

###Both Boolean and Query Does the same job, as we can see in above two queries
###Query is best for readability

#Problem
###Find IT employees with salary > 65000.

###Boolean Method

```python
df[(df["Department"]=="IT") & (df["Salary"]>65000)]
```

###Query Method

```python
df.query("Department == 'IT' and Salary > 65000")
```

#Some Professional Tricks
##Trick 1 — Selecting Rows & Columns Dynamically (Using lists with loc)

```python
df
```

###Suppose you want to select only specific rows

```python
rows = [0,3,4]
columns = ['Name', 'Department']
df.loc[rows,columns]
```

##Trick 2 — Conditional Row & Column Selection with loc

###You can combine Boolean filtering and column selection in one step, saving multiple lines of code.

###Example:

Get Name and Salary for employees with Salary > 70000:

Instead of writing two steps (filter + select columns), you do it in one line.

```python
df.loc[df["Salary"]>70000, ["Name", "Salary"]]
```

##Trick 3 — Selecting Rows & Columns by Position + Slicing

```python
df.iloc[[0,2],[0,3]]
```

###Selecting all rows but only last two columns

```python
df.iloc[:,-2:]
```

###You can combine lists, slices, and Boolean masks in both

```python
df.loc[df["Salary"]>7000, 'Department'] = 'Management'
```
