#Topic 1: GroupBy Fundamentals
###1. What is GroupBy?
####GroupBy is used to split data into groups based on one or more columns, apply operations on each group, and then combine the results.
###Split → Apply → Combine

####Split → Divide data into groups

####Apply → Perform operation (sum, mean, etc.)

####Combine → Merge results

```python
import pandas as pd

data = {
    "City": ["Lahore", "Karachi", "Lahore", "Islamabad", "Karachi", "Lahore"],
    "Product": ["Laptop", "Mobile", "Tablet", "Laptop", "Tablet", "Mobile"],
    "Sales": [100000, 50000, 30000, 70000, 20000, 60000],
    "Profit": [20000, 10000, 5000, 15000, 3000, 12000]
}
df = pd.DataFrame(data)
df
```

```python
df.groupby('City')
```

####To get meaningful results, we must apply an aggregation function.

```python
df.groupby("City")['Sales'].sum()
```

####Total Sales per City

###Grouping by Multiple Columns

```python
df.groupby(["City", "Product"])["Sales"].sum()
```

###Understanding the Output (VERY IMPORTANT)

####Output is a Series with hierarchical index (MultiIndex)

###Example:

####City       Product
####Lahore     Laptop     100000
####           Mobile      60000


####This is called a MultiIndex (nested index).

##As_Index Parameter
####By default, groupby sets grouped column as index.

```python
df.groupby("City", as_index= False)['Sales'].sum()
```

```python
df.groupby("City")['Sales'].sum()
```

##Iterating Through Groups
####We can loop through each group for deeper understanding.

```python
for city, group in df.groupby("City"):
  print(city)
  print(group)
```

- GroupBy splits data into groups
- Requires aggregation to produce results
- Can group by one or multiple columns
- Output often uses MultiIndex
- as_index=False makes output cleaner

#Topic 2: Basic Aggregation Functions
###1. What are Aggregation Functions?
####Aggregation functions are used to summarize data by performing calculations on groups.

####They reduce multiple values into a single value.
###Examples:

####Many sales → Total sales

####Many salaries → Average salary

###Why Aggregation is important
####Aggregation helps in extracting meaningful insights from data, such as:
- Total revenue
- Average profit
- Maximum sales
- Minimum performance

###Common aggregation Functions
####The most important aggregation functions are:
- sum() → Total
- mean() → Average
- count() → Number of values
- min() → Smallest value
- max() → Largest value
- median() → Middle value

##Using Aggregation with GroupBy
###Sum()
Total Sales per City

```python
df.groupby("City")["Sales"].sum()
```

###Mean (Average)

```python
df.groupby("City")['Profit'].mean()
```

###Count

```python
df.groupby("City")['Product'].count()
```

###Min()

```python
df.groupby("City")['Sales'].min()
```

###Max()

```python
df.groupby("City")['Sales'].max()
```

###Median() Middle Value (useful when data has outliers)

```python
df.groupby("City")['Sales'].median()
```

###Applying Aggregation Without GroupBy
####Aggregation can also be applied to entire columns.

```python
df.groupby("City").sum()
```

```python
df.groupby("Sales").sum()
```

###Multiple Columns Aggregation
####We can apply aggregation on multiple columns at once.

```python
df.groupby("City")[['Sales','Profit']].sum()
```

###Difference Between count() and size()
####count() → counts non-null values
####size() → counts all rows (including nulls)

```python
df.groupby("City")["Sales"].count()
```

```python
df.groupby("City")["Sales"].size()
```

##Numeric Columns Only Behavior
####Aggregation functions automatically ignore non-numeric columns
####(except count()).

###Chaining with Sorting (Real Insight)
###Instantly shows:

####Best performing city

####Worst performing city

```python
df.groupby("City")['Sales'].sum().sort_values(ascending=False)
```

###Output Type
###Aggregation usually returns:
- Series (single column)
- DataFrame (multiple columns)

####Aggregation = “Summarizing data into insights”
- Aggregation reduces multiple values to one value
- Works with groupby to summarize data
- sum, mean, count, min, max are most important
- median is useful for outliers
- Sorting makes results meaningful

#Topic 3: Multiple Aggregations (.agg())

##1. What is Multiple Aggregation?
####Multiple aggregation means applying more than one aggregation function
on grouped data at the same time.

###Instead of:

####Only sum

####Only mean

###You can get:

####sum + mean + max → together

##2. Using .agg() with Single Column
####We can apply multiple functions to a single column using .agg().

```python
df.groupby("City")['Sales'].agg(['sum','mean','min'])
```

####Output:

Total sales

Average sales

Maximum sales
###.agg() accepts a list of functions

##3. Using .agg() with Multiple Columns
####We can apply multiple aggregations on multiple columns.
###Output Note
####This creates MultiIndex columns (nested column names).

```python
df.groupby("City")[['Sales','Profit']].agg(['sum','mean'])
```

##4. Different Functions for Different Columns
####We can assign different aggregation functions to different columns.

```python
df.groupby("City").agg({
    "Sales": "sum",
    "Profit": "mean"})
```

##5. List of Functions in .agg()
####Common functions used in .agg():

- "sum"
- "mean"
- "count"
- "min"
- "max"
- "median"
- "std" (standard deviation)

##6. Named Aggregations
####Named aggregation allows us to give custom column names to results.

```python
df.groupby("City").agg(
    Total_Sales = ("Sales", "sum"),
    Avg_Profit = ("Profit", "mean")
        )
```

##7. Output Structure & Understanding
###.agg() can return:
- DataFrame with multiple columns
- MultiIndex columns (if not named properly)

###Example:


```python
df.groupby("City")["Sales"].agg(["sum", "mean"])
```

##8. Combine with Sorting


```python
df.groupby("City").agg(
    Total_Sales= ("Sales", "sum"),
    Avg_Profit = ("Profit", "mean")
).sort_values(by= "Total_Sales", ascending = False)
```

###ALL Codes at one place
`df.groupby("City")[["Sales","Profit"]].agg("sum","mean").sort_values(by= "Sales", ascending=False)`

`df.groupby("City").agg({"Sales":"sum","Profit":"mean"}).sort_values(by= "Profit", ascending=False)`

`df.groupby("City").agg(Total_Sales= ("Sales","sum"), Avg_profit = ("Profit", "mean")).sort_values(by= "Avg_profit", ascending=False)`

##9. When to Use .agg() vs Normal Aggregation
###Use simple aggregation when:
- Only one function is needed

###Use .agg() when:
- Multiple functions are needed
- Different columns need different operations

###.agg() = “Multiple summaries in one step”
- .agg() allows multiple aggregations at once
- Can be used on single or multiple columns
- Can apply different functions to different columns
- Named aggregation makes output clean and professional

#Topic 4: Grouping by Multiple Columns
##1. What is Multi-Column Grouping?
####Grouping by multiple columns means splitting data based on more than one category.

####- Instead of grouping by just 'City' → group by 'City' AND 'Product'.
####- Useful for business insights like:
    - Total sales per product in each city
    - Average profit per department in each region

##2. Syntax for Multiple Columns
`df.groupby(["City", "Product"])`

##3. Aggregation on Multiple Columns

```python
df.groupby(["City", "Product"])[["Sales", "Profit"]].sum()
```

##4. MultiIndex Output (Understanding the Structure)
###Grouping by multiple columns produces a MultiIndex:

###Example:

City       Product

Karachi    Mobile
           Tablet

Lahore     Laptop
           Mobile

###City → first level of index

###Product → second level of index

###Values → aggregated numbers

##5. Accessing Specific Groups
###Using .loc on MultiIndex

```python
grouped = df.groupby(["City", "Product"])[["Sales", "Profit"]].sum()
grouped
```

```python
grouped.loc[("Islamabad","Laptop")]
```

```python
grouped.loc['Karachi','Tablet']
```

###using .xs(Cross-Section)

```python
grouped
```

```python
grouped.xs('Karachi', level='City')
```

```python
grouped.xs("Laptop", level = 'Product')
```

##6. Sorting Multi-Column Groups
####Sort by first column (City) and then Sales

```python
df.groupby(['City', "Product"])['Sales'].sum().sort_values(ascending= False)
```

###Sort by index levels

```python
grouped
```

```python
grouped.sort_index(level="City")
```

```python
grouped.sort_index(level= ("City", "Product"))
```

##7. Resetting Index for Cleaner Output

```python
df.groupby(["City","Product"])[['Sales',"Profit"]].sum().reset_index()
```

#Topic 5: Sorting Aggregated Data

##1. Why Sorting is Important
####Aggregated/grouped data is useful, but raw output is often unordered.

###Sorting helps:
- Identify top/bottom performers
- Quickly find insights
- Make reports more readable

##2. Sorting a Series
####Example: Grouped Sales by City

```python
grouped_sales= df.groupby("City")["Sales"].sum()
grouped_sales
```

###Output unordered
###Now Sort by values

```python
grouped_sales.sort_values(ascending=False)
```

##3. Sorting a DataFrame

####If you reset index after grouping:

```python
grouped_sales.reset_index()
```

###Sort by a column

```python
df.groupby(["City","Product"])[["Sales","Profit"]].sum().sort_values(by = "Profit", ascending=False).reset_index()
```

##4. Sorting by Index

####If grouped output is a Series with MultiIndex:

```python
grouped = df.groupby(["City", "Product"])["Sales"].sum()
grouped
```

```python
grouped.sort_index(level=['City','Product'])
```

##5. Ascending vs Descending
### Descending (default False)
`grouped.sort_values(ascending=False)`

### Ascending (default True)
`grouped.sort_values(ascending=True)`

##6. Sorting MultiIndex Groups
####Sort by values in MultiIndex

```python
grouped= df.groupby(["City","Product"])['Sales'].sum().sort_values(ascending=False)
grouped
```

###Sort by Index

```python
grouped.sort_index(level= ("City","Product"),ascending=[False,False])
```

##7. Reset Index for Clean Output

```python
df.groupby(["City", "Product"]).agg(
    Total_values= ("Sales", "sum"),
    Avg_Profit = ("Profit", "mean")
).sort_values(by= ["Total_values", 'Avg_Profit'] ,ascending= [True, False]).sort_index(level=["City", "Product"], ascending= [True,False]).reset_index()
```

##Key Points
- Sorting is essential for readable aggregated data
- Use sort_values() to sort by aggregated values
- Use sort_index() to sort by index (single or MultiIndex)
- Chaining with groupby makes analysis clean & professional
- Always reset index for portfolio-ready output

###Chaining groupby, aggregation, and sorting in one line creates
###clean, professional, and report-ready analysis.

##Just three operations to keep in mind:


1.   sort_values(ascending= False)
2.   sort_index(level= ("City", "Product"), ascending = [True,False])
3.   reset_index()






#Topic 6: Counting & Frequencies (value_counts & .count())

##1. Difference Between count() and value_counts()
- `count()` → Counts non-null values in a column (or group)
- `value_counts()` → Counts **frequency of each unique value**

##2. Using value_counts() for a Single Column
####Instantly shows how many rows belong to each city

```python
df['City'].value_counts()
```

##3. Sorting value_counts()

####By default, value_counts() sorts descending by frequency:

```python
df['City'].value_counts()
```

####Ascending sort:

```python
df['City'].value_counts(ascending=True)
```

##4. Normalizing value_counts() (Percentages)


```python
df['City'].value_counts(normalize=True)
```

##5. Handling NaN Values

```python
df.groupby('City')['Product'].value_counts(dropna=False)
```

Includes NaN values in counts

Helps detect missing categories

##6. GroupBy + count()
####We can count rows per group using GroupBy

```python
df.groupby("City")['Sales'].count()
```

##7. GroupBy + value_counts() (Advanced)

####Sometimes you want distribution of one column within groups:

```python
df.groupby("City")["Product"].value_counts()
```

##Key Points
- count() → counts non-null rows
- value_counts() → counts frequency of unique values
- normalize=True → gives percentages
- dropna=False → includes missing values
- Combine with groupby for segmented frequencies
- Essential for EDA & quick portfolio analysis

#Topic 7: Ranking in Pandas

##1. What is Ranking?
####Ranking assigns a rank or position to each value in a dataset based on its magnitude.

###Example:
- Highest sales → rank 1
- Lowest sales → rank last

##2. Using .rank() on a Series

```python
df['Sales'].rank()
```

##3. Ranking Methods
###Methods:

average → default, average rank for ties

min → all ties get minimum rank

max → all ties get maximum rank

first → ranks according to order in data

dense → like min but no gaps

```python
import pandas as pd

data = {
    "City": ["Lahore", "Karachi", "Lahore", "Islamabad", "Karachi", "Lahore"],
    "Product": ["Laptop", "Mobile", "Tablet", "Laptop", "Tablet", "Mobile"],
    "Sales": [100000, 50000, 70000, 70000, 20000, 100000],
    "Profit": [20000, 10000, 5000, 15000, 3000, 12000]
}
df1 = pd.DataFrame(data)
df1
```

```python
df1["Sales"].rank(method="min")
```

```python
df1["Sales"].rank(method="max")
```

```python
df1["Sales"].rank(method="first")
```

```python
df1["Sales"].rank(method="dense")
```

##4. Descending vs Ascending
#### Highest value gets rank 1

```python
df1['Sales'].rank(ascending=False)
```

##5. Ranking Within Groups (GroupBy + rank)

```python
df1.groupby("City")["Sales"].rank(ascending=False)
```

##6. Resetting Index After Ranking

```python
ranked_df = df1.groupby("City")["Sales"].rank(ascending=False).reset_index()
```

```python
ranked_df
```

##Key Points
- `.rank()` assigns order to values
- Can rank ascending or descending
- Several methods to handle ties: average, min, max, first, dense
- Works within groups using GroupBy
- Essential for top-n analysis and leaderboards
- Reset index for clean portfolio outputs

#Topic 10: Filtering Groups (.filter())
##1. What is Group Filtering?
####Group filtering is the process of **removing or keeping groups** based on a condition applied to the aggregated group.

###Example:
- Only include cities where total sales > 50,000
- Only include products with average profit > 10,000

##2. Syntax of .filter()
`df.groupby("City").filter(func)`

####func → a function that takes a grouped DataFrame and returns True (keep) or False (drop)

##3. Filtering by Aggregated Value
####Example: Keep cities with total sales > 100,000

```python
df.groupby("City").filter(lambda x: x['Sales'].sum()> 100000)
```

##4. Filtering by Multiple Conditions
####Combine conditions with & (and) or | (or)

####Flexible for complex business rules

```python
df.groupby("City").filter(lambda x: (x['Sales'].sum()> 100000) & (x['Profit'].mean()> 5000))
```

##5. Using Lambda Functions
#### Keep groups where max sale > 80,000

```python
df.groupby("City").filter(lambda x: x['Sales'].max()> 80000)
```

##6. Filter and Keep Only Specific Groups
#### Keep only "Lahore" group

```python
df.groupby("City").filter(lambda x: x.name == 'Lahore')
```

####.name → gives the group name (City in this case)

##7. Combining with GroupBy + Aggregation

####You can chain aggregation + filtering:


####Filter first → then select relevant columns

```python
df.groupby("City").filter(lambda x: x["Sales"].sum() > 100000)[["City", "Product", "Sales"]]
```

##Key Points
- .filter() lets you keep or remove groups based on a condition
- Works on grouped DataFrames
- Use lambda functions for dynamic conditions
- Can combine multiple conditions
- Essential for focusing on meaningful data in analysis and reporting
