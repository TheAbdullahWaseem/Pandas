#2. Pandas Data Selection and Indexing
In this notebook, we explore data selection and indexing in Pandas, which are essential skills for extracting and filtering specific information from a dataset.

The following key concepts are covered:

**Selecting Columns** – Accessing one or multiple columns from a DataFrame using bracket notation.

**Selecting Rows** – Retrieving rows using index labels or positions.

**loc Indexer** – Label-based selection used to access rows and columns by their names or labels.

**iloc Indexer** – Position-based selection used to access rows and columns by integer index.

**Slicing Data** – Selecting ranges of rows or columns using slicing syntax.

**Boolean Filtering** – Filtering rows based on logical conditions (e.g., salary greater than a certain value).

**Conditional Filtering** – Combining multiple conditions using logical operators such as & (AND) and | (OR).

**Querying Data** – Using the query() method to filter data using readable conditional expressions.

```python
import pandas as pd
```

#Practice Questions
###1. Select only Name and Salary columns

```python
data = {
    "Name": ["Ali", "Amir", "Sara", "Hina", "Fahad"],
    "Age": [23, 25, 22, 24, 26],
    "Department": ["IT", "Finance", "IT", "HR", "Finance"],
    "Salary": [70000, 80000, 65000, 60000, 90000]
}
df = pd.DataFrame(data)
df
```

```python
df[['Name','Salary']]
```

###2. Select rows 1 to 3 using loc

```python
df.loc[1:3]
```

###3. Select rows 1 to 3 using iloc

```python
df.iloc[1:4]
```

###4. Find employees with:
###Salary > 70000

```python
df[df['Salary']>70000]
```

###5. Find employees who work in:
###IT department

```python
df.query("Department == 'IT'")
```

###6. Find employees with:
###Age > 23 AND Salary > 65000

```python
df[(df["Age"]>23) & (df['Salary']> 65000)]
```

###7. Find employees with:
###Age > 23 AND Salary > 65000 using Query

```python
df.query("Age > 23 and Salary > 65000")
```

###8. Select Name and Salary of employees where:

###Salary > 70000

```python
df.loc[df['Salary']> 70000, ['Name', 'Salary']]
```
