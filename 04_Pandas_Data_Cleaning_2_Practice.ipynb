#Pandas Data Cleaning 2 Practice

```python
import pandas as pd
```

```python
data = {
    "Name": ["Ali", "Sara", "Ali", "Hina", "Omar", "Sara"],
    "Age": ["25", "30", "25", None, "28", "30"],
    "Salary": ["50000", "60000", "50000", "55000", "not available", "60000"],
    "JoinDate": ["2023-01-15", "2023/02/20", "2023-01-15", "15-03-2023", None, "2023/02/20"],
    "Department": ["IT", "HR", "IT", "Finance", "IT", "HR"]
}
df = pd.DataFrame(data)
df
```

```python
df.info()
```

##Question 1 — Detect duplicates

###Find all rows that are duplicates considering all columns.

```python
df.duplicated()
```

##Question 2 — Detect partial duplicates

###Find duplicates based only on Name and Department.

```python
df.duplicated(subset= ['Name','Department'])
```

##Question 3 — Remove duplicates

###Remove duplicate rows based on Name and Department, keeping only the first occurrence.

```python
df.drop_duplicates(subset= ['Name','Department'],keep= 'first')
```

##Question 4 — Safe numeric conversion

###Convert the Salary column to numeric.
####Invalid values like "not available" should become NaN.

```python
df['Salary']= pd.to_numeric(df['Salary'], errors= 'coerce')
```

##Question 5 — Convert multiple columns

###Convert Age and Salary columns to float type in one command.

```python
df[['Age','Salary']].astype(float)
```

##Question 6 — Datetime conversion

###Convert the JoinDate column to datetime so that all valid dates are correctly parsed and invalid ones become NaT.

```python
df['JoinDate'] = df['JoinDate'].str.replace('/','-')
```

```python
df['JoinDate']
```

```python
pd.to_datetime(df['JoinDate'], errors= 'coerce', dayfirst= True)
```

##To change the date at index 4(15-03-2023)
###We perform the following method

```python
mask = df['JoinDate'].isna()
mask
```

```python
df.loc[mask, 'JoinDate'] = pd.to_datetime(df.loc[mask, 'JoinDate'], format= '%Y-%m-%d', errors= 'coerce')
```
