#3. Pandas Data Cleaning 1 Practice

```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Name": ["Ali", "Sara", "Ali", "Hina", "Omar", "Sara"],
    "Age": [23, np.nan, 23, 25, np.nan, 28],
    "Salary": [50000, 60000, 50000, np.nan, 70000, 60000],
    "Department": ["IT", "HR", "IT", "Finance", "IT", "HR"]
}
df = pd.DataFrame(data)
df
```

##PART 1 — Missing Data
###Question 1 (Concept Test)

###Write a command that shows which cells in the dataset contain missing values.

```python
df.info()
```

```python
df.isna()
```

##Question 2 (Counting Missing Data)

###Find how many missing values exist in each column.

###Expected type of result:

**Age       ?**

**Salary    ?**

```python
df[['Age','Salary']].isna().sum()
```

##Question 3

###Write a command that removes rows where Age is missing, but keeps rows where Salary is missing.

```python
df.dropna(subset=['Age'])
```

##Question 4 (Logic Test)

###Remove rows only if both Age and Salary are missing.

```python
df.dropna(subset=['Age','Salary'],how='all')
```

##Question 5 (Data Imputation)

###Replace missing values in Age with the average age.

```python
df['Age'].fillna(df['Age'].mean())
```

##Question 6 (Selective Fill)

###Replace missing Salary with 50000.

```python
df['Salary'].fillna(50000)
```

##PART 2 — Duplicates
###Question 7 (Detect Duplicates)

###Write a command that identifies duplicate rows.

```python
df.duplicated()
```

##Question 8 (Remove Duplicates)

###Write a command that removes duplicate rows and keeps only unique rows.

```python
df.drop_duplicates()
```

##Question 9 (Advanced Duplicate Logic)

###Remove duplicates based only on Name, keeping the first occurrence.

```python
df.drop_duplicates(subset=['Name'],keep='first')# you can use keep = 'last' or False
```

##Question 10 (Professional Trick)

###Remove duplicates based on Name and Department together.

```python
df.drop_duplicates(subset=['Name','Department'])
```
