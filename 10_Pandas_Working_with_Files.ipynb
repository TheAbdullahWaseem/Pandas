##5. Writing Large Files in Chunks

###1. What are Chunks?

####A chunk is basically a smaller part of a big dataset.

####Think of a huge Excel file with millions of rows. Loading it all into memory at once can crash your computer, because Pandas tries to read everything at once.

###2. Why Use Chunks?
####1. Memory efficiency: Your RAM isn’t overloaded.
####2. Speed: Process smaller parts quickly instead of waiting for the entire file.
####3. Streaming large files: Some files are too big to load at once (millions of rows).
####4. Incremental processing: You can calculate aggregates, filter, or clean chunk by chunk.

###3. How to Use Chunks in Pandas

#####There are two main cases:

###A. Reading large CSV files in chunks

####Use chunksize in pd.read_csv():

```python
chunk_size = 200
for chunk in pd.read_csv('tested.csv', chunksize= chunk_size):
  print(chunk.shape)
```

####pd.read_csv() doesn’t load the whole file, only 200 rows at a time.
####Each chunk is a DataFrame you can process.

###B. Writing large CSV files in chunks

####Sometimes the DataFrame is already in memory, but it’s huge.

####Writing all rows at once may consume too much memory.
Solution: split DataFrame into chunks and write incrementally.

```python
chunk_size = 200  # rows per chunk

for i, start in enumerate(range(0, len(df), chunk_size)):
    df.iloc[start:start+chunk_size].to_csv(
        'titanic_large_chunks.csv',
        mode='a',          # append to file
        index=False,
        header=(i==0),     # write header only for first chunk
        na_rep='MISSING'
    )
```
