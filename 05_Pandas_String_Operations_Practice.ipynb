#Pandas String Operation Practice

```python
import pandas as pd

data = {
    "customer_id": [101,102,103,104,105,106,107,108],

    "name": [
        "  ali khan",
        "SARA AHMED ",
        " ahMeD RAZA ",
        "Fatima Noor",
        "  usman ali ",
        "Ayesha Malik ",
        "bilal  khan",
        " HAMZA SHAH"
    ],

    "email": [
        "ali.khan@gmail.com ",
        "SARA@YAHOO.COM",
        " ahmed123@gmail.com",
        "fatima_noor@hotmail.com ",
        "usman.ali@gmail.com",
        "ayesha.malik@yahoo.com ",
        "bilal-khan@gmail.com",
        "hamza.shah@outlook.com"
    ],

    "city": [
        " new-york ",
        "LOS-ANGELES",
        " chicago ",
        "Houston",
        "phoenix ",
        " SAN-ANTONIO",
        "san-diego",
        " Dallas "
    ],

    "job_title": [
        "data analyst",
        "DATA SCIENTIST",
        "machine learning engineer",
        "software developer",
        "data engineer",
        "business analyst",
        "AI researcher",
        "data ANALYST"
    ]
}

df = pd.DataFrame(data)
df
```

```python
df.info()
```

#Question 1 — Full Name Cleaning Pipeline

###Clean the name column so that:

####"  ali khan" → "Ali Khan"
####"bilal  khan" → "Bilal Khan"

###Requirements:

####remove extra spaces

####ensure only one space between words

####convert to title case

```python
df['name'].str.strip().str.replace(r"\s+"," ",regex=True).str.title()
```

#Question 2 — Email Standardization

###Clean the email column so that:

####"SARA@YAHOO.COM" → "sara@yahoo.com"
####" ali.khan@gmail.com " → "ali.khan@gmail.com"

###Requirements:

####remove spaces

####convert to lowercase

```python
df['email'].str.strip().str.lower()
```

#Question 3 — Extract Email Domain

####Create a new column called domain.

###Example:

####ali.khan@gmail.com → gmail.com

```python
df['domain']= df['email'].str.strip().str.split("@").str[1].str.lower()
df['domain']
```

#Question 4 — Gmail Users Filter

###Create a new dataframe containing only Gmail users.

```python
gmail_users= df[df['email'].str.rstrip().str.endswith("@gmail.com")]
gmail_users
```

#Question 5 — City Cleaning

###Clean the city column so that:

####" new-york " → "New York"
####"LOS-ANGELES" → "Los Angeles"

###Requirements:

####remove spaces

####replace - with space

####convert to title case

```python
df['city'].str.strip().str.replace("-", " ").str.title()
```

#Question 6 — Detect Double Spaces in Names

####Find rows where the name contains multiple spaces between words.

###Example problem:

####"bilal  khan"

```python
df['name'].str.strip().str.contains(r"\s{2,}",regex=True)
```

#Question 7 — Extract First Name

###Create a new column:

####first_name

###Example:

####Ali Khan → Ali
####Sara Ahmed → Sara

```python
df['first_name']= df['name'].str.strip().str.replace(r"\s+"," ",regex=True).str.split(" ", expand = True)[0].str.title()
df['first_name']
```

#Question 8 — Extract Last Name

###Create another column:

####last_name

###Example:

####Ali Khan → Khan
####Sara Ahmed → Ahmed

```python
df['last_name']= df['name'].str.strip().str.replace(r"\s+"," ",regex=True).str.split(" ", expand = True)[1].str.title()
df['last_name']
```

#Question 9 — Length of Names

####Create a column:

###name_length

####This should store the length of each cleaned name.

```python
df['name_length'] = df['name'].str.strip().str.len()
df['name_length']
```

#Question 10 — Job Title Filtering

####Find rows where job title contains the word "data".

####The search should not fail because of capitalization differences.

```python
df['job_title'].str.contains("data",case=False)
```
