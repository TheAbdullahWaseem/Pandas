#1. Missing Data
What is Missing Data?

Missing data occurs when values are absent from a dataset.

In Pandas, missing values are usually represented as:

**NaN**

```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Name": ["Ali", "Sara", "Amir", "Hina","Ali","Sara"],
    "Age": [23, np.nan, 25, 24,23,22],
    "Salary": [70000, 65000, np.nan, 60000,70000,65000]
}
df = pd.DataFrame(data)
df
```

Here:

###Sara’s **Age** is missing

###Amir’s **Salary** is missing

##1.1 Detecting Missing Values
###The **isna()** function identifies missing values.

```python
df.isna()
```

**True** = Missing value

**False** = Value exists

##Counting Missing Values

Often we want to know how many missing values exist in each column.

```python
df.isna().sum()
```

###Age column has 1 missing value

###Salary column has 1 missing value

##1.2 Detecting Non-Missing Values — notna()

**notna()** is the opposite of **isna()**.

It returns **True for existing values** and **False for missing values.**

```python
df.notna()
```

##1.3 Removing Missing Values — dropna()

###**dropna()** removes rows or columns containing missing values.

```python
df.dropna()
```

```python
df
```

##Dropping Columns with Missing Values

###If a column contains NaN, it will be removed.

axis = 1 → columns

axis = 0 → rows

```python
df.dropna(axis=1)
```

##Dropping Rows Only if ALL Values Are Missing

```python
df.dropna(how='all')
```

###Dropping Rows Based on Specific Columns

```python
df.dropna(subset=['Salary','Age'])
```

```python
df
```

##1.4 Filling Missing Values — fillna()

All missing values become **0**.

```python
df.fillna(0)
```

##Fill with Specific Value

```python
df['Salary'].fillna(80000)
```

```python
mean = np.mean(df['Salary'])
mean
```

##Fill with Mean Value
This replaces missing values with the **average.**

```python
df['Salary'].fillna(mean)
#OR
df['Salary'].fillna(df['Salary'].mean())
```

##Forward Fill Method
fill missing value using the previous row's value

```python
df.fillna(method='ffill')
```

```python
df['Salary'].fillna(method='ffill')
```

##Backward Fill Method
Fill missing value using next row's value

```python
df['Age'].fillna(method='bfill')
```

```python
df.fillna(method='bfill')
```

#2. Duplicate Data

###Sometimes datasets contain repeated rows.

Example:

Name	  Age

Ali	    23

Sara	  24

Ali	    23

The third row is a duplicate.

##Detecting Duplicates
###duplicated()

True mean the row is duplicated

```python
df.duplicated()
```

##Removing Duplicates
###drop_duplicates()
Duplicate rows will be removed.

```python
df.drop_duplicates()
```

##Removing Duplicates Based on Specific Column
This removes duplicate Salaries.

```python
df.drop_duplicates(subset=['Salary'])
```
