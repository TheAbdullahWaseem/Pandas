#Pandas String Operations
##Overview

In real-world datasets, text data is often **messy and inconsistent**. Names may have extra spaces, inconsistent capitalization, unwanted characters, or combined fields that need to be separated.

To perform reliable analysis, this text must be **cleaned and standardized**.

In this lesson we will learn how to clean and manipulate text data in pandas using the **.str accessor**, which allows vectorized string operations on an entire column.

##Topics Covered:

**1. Introduction to String Operations in Pandas**

**4. str.upper() – convert text to uppercase**

**7. str.contains() – search for patterns in text**

**8. str.split() – split strings into multiple parts**

**9. str.len()**

**10. str.startswith()**

**11. str.endswith()**

#1. Why String Cleaning Is Important

##Real datasets often contain problems like:

###Problem	Example
####Inconsistent capitalization	**Ali, ALI, ali**
####Extra spaces	**" Ahmed "**
####Mixed formatting	**"New-York" vs "New York"**
####Combined fields	**"Ali Khan"**

If this data is not cleaned, analysis results will be incorrect or misleading.


#2. The .str Accessor

In pandas, string operations are performed using the .str accessor.

Example:

####df["name"].str.lower()


###Explanation:

####df["name"] → select the column

####.str → access string methods

####.lower() → apply lowercase conversion

```python
df['name'].str.lower()
```

#3. str.lower()
###Purpose

####Convert all text in a column to lowercase.

```python
df['city'].str.lower()
```

#4. str.upper()
###Purpose

####Convert text to uppercase.

```python
df['name'].str.upper()
```

#5. str.strip()
###Purpose

####Remove extra spaces from the beginning and end of strings.

```python
df['name'].str.strip()
```

##Related Methods

###Remove only left spaces:

####**str.lstrip()**


###Remove only right spaces:

####**str.rstrip()**

```python
df['name'].str.lstrip()
```

```python
df['name'].str.rstrip()
```

```python
df['name'] = df['name'].str.strip()
```

###Can remove specific characters:

```python
df['email'].str.strip("$@")
```

#6. str.replace()
###Purpose

####Replace characters or patterns in strings.

```python
df['city']
```

```python
df['city'].str.replace("-"," ")
```

#7. str.contains()
##Purpose

####Check if a string contains a specific pattern.

####Returns True or False.

```python
df['email'].str.contains('gmail')
```

#8. str.split()
##Purpose

####Split text into multiple parts.

```python
df['name'].str.split(" ")
```

###Splitting Into Columns

```python
df['name'] = df['name'].str.strip()
```

```python
df['name'] = df['name'].str.replace("  ", " ")
```

```python
df['name'].str.split(" ",expand=True)
```

#9. str.len()

###Find the length of strings.

```python
df['name'].str.len()
```

#10. str.startswith()

###Check if string starts with a pattern.

```python
df['email'].str.startswith('ali')
```

#11. str.endswith()
###Check if string ends with a pattern.

```python
df['email'].str.endswith('.com')
```

#12. str.title()

###Convert text to title case.

```python
df['name'].str.title()
```

```python
df['city'].str.title()
```

#Real-World String Cleaning Workflow

Example messy dataset:

name	city
" ali "	"new-york"
"SARA "	"los-angeles"

Cleaning workflow:

`df["name"] = df["name"].str.strip().str.lower().str.title()`

`df["city"] = df["city"].str.replace("-", " ").str.title()`
